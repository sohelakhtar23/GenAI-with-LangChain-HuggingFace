```python
from dotenv import load_dotenv
import os   
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
if not groq_api_key:
    raise ValueError("GROQ_API_KEY environment variable is not set. Please set it to your Groq API key.")
```

```python
from langchain_core.messages import AIMessage, HumanMessage
from langchain_groq import ChatGroq
groq_model = ChatGroq(
    model="Gemma2-9b-It",
    groq_api_key=groq_api_key,
)

groq_model.invoke([
    HumanMessage(content="I am from Nepal. The weather is very hot today. What is the weather like in San Francisco?"),  
])
```

```python
groq_model.invoke([
    HumanMessage(content="I am from Nepal. The weather is very hot today. What is the weather like in San Francisco?"),  
    AIMessage(content="I have no access to real-time weather data, but I can tell you that San Francisco typically has a mild climate with cool summers and fog. You might want to check a reliable weather website or app for the current conditions in San Francisco."),
    HumanMessage(content="Where am I from?"),
])
```

## Message history

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store ={}
def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_history = RunnableWithMessageHistory(
    runnable=groq_model,
    get_session_history=get_session_history,
)
```

```python
config1 = {"configurable": {"session_id": "session_1"}}

with_history.invoke(
    [HumanMessage(content="I am from Nepal. The weather is very hot today. What is the weather like in San Francisco?"),],
    config= config1
)
```

```python
with_history.invoke(
    [HumanMessage(content="Where am I from?")],
    config= config1
)
```

```python
# Changing the session_id to see how it affects the history
with_history.invoke(
    [HumanMessage(content="Where am I from?")],
    config= {"configurable": {"session_id": "new_session"}}
)
```
